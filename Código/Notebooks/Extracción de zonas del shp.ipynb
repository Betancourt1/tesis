```python
import shapefile
import csv
import os

# --- Configuración ---
# Ruta a tu archivo Shapefile
ruta_shapefile = r"C:\Users\fbetancourt\Documents\vwvlimitecentralidad.shp"
# Codificación del archivo .dbf (ajusta si 'latin1' no funcionó antes, prueba 'cp1252')
codificacion_dbf = 'latin1'
# Nombres para los archivos CSV de salida
nombre_csv_coordenadas = r"C:\Users\fbetancourt\Documents\GitHub\Tesis\Datasets\shapes_imeplan\coordenadas_con_indice.csv"
nombre_csv_atributos = r"C:\Users\fbetancourt\Documents\GitHub\Tesis\Datasets\shapes_imeplan\atributos_centralidad.csv"
# Carpeta de salida para los CSV (se guardarán en la misma carpeta del script por defecto)
carpeta_salida = os.path.dirname(__file__) if '__file__' in locals() else '.' # Directorio del script o directorio actual
ruta_csv_coordenadas = os.path.join(carpeta_salida, nombre_csv_coordenadas)
ruta_csv_atributos = os.path.join(carpeta_salida, nombre_csv_atributos)

# Nombres de los campos a extraer (basado en la info previa)
campo_indice = 'Identifica'
campo_nombre = 'Nombre_de_'
campo_tipo = 'Tipo_de_ce'
# --------------------

try:
    # Abre el archivo Shapefile en modo lectura, especificando la codificación
    sf = shapefile.Reader(ruta_shapefile, encoding=codificacion_dbf)
    print(f"Leyendo Shapefile: {ruta_shapefile}")
    print(f"Tipo: {sf.shapeTypeName}, Codificación DBF: {codificacion_dbf}")

    # --- Obtener índices de los campos necesarios ---
    campos = sf.fields[1:] # Omitimos 'DeletionFlag'
    nombres_campos = [field[0] for field in campos]

    try:
        indice_id = nombres_campos.index(campo_indice)
        indice_nombre = nombres_campos.index(campo_nombre)
        indice_tipo = nombres_campos.index(campo_tipo)
        print(f"Índices encontrados -> ID: {indice_id}, Nombre: {indice_nombre}, Tipo: {indice_tipo}")
    except ValueError as e:
        print(f"Error: No se encontró uno de los campos requeridos ({e}) en {nombres_campos}")
        exit() # Salir si no se encuentran los campos necesarios

    # --- Procesar y escribir CSV 1: Coordenadas con Índice ---
    print(f"\nCreando CSV 1: {ruta_csv_coordenadas}")
    # --- Procesar y escribir CSV 1: Coordenadas con Índice ---
    print(f"\nCreando CSV 1: {ruta_csv_coordenadas}")
    with open(ruta_csv_coordenadas, 'w', newline='', encoding='utf-8') as csvfile1:
        writer1 = csv.writer(csvfile1)
        # Escribir encabezado modificado
        writer1.writerow(['indice_centralidad', 'orden_punto', 'longitud', 'latitud'])

        # Iterar sobre shapes y records
        for shapeRecord in sf.iterShapeRecords():
            shape = shapeRecord.shape
            record = shapeRecord.record

            # Obtener el índice de la centralidad
            indice_centralidad = record[indice_id]

            # Inicializar el contador de orden de puntos para este polígono
            orden_punto = 0
            # Iterar sobre cada punto (coordenada) de la geometría
            for punto in shape.points:
                longitud = punto[0]
                latitud = punto[1]
                # Escribir fila por cada punto, incluyendo el orden
                writer1.writerow([indice_centralidad, orden_punto, longitud, latitud])
                orden_punto += 1 # Incrementar el orden del punto

    print("CSV 1 creado exitosamente.")

    # --- Procesar y escribir CSV 2: Atributos de Centralidad ---
    print(f"\nCreando CSV 2: {ruta_csv_atributos}")
    with open(ruta_csv_atributos, 'w', newline='', encoding='utf-8') as csvfile2:
        writer2 = csv.writer(csvfile2)
        # Escribir encabezado
        writer2.writerow(['indice_centralidad', 'nombre', 'tipo'])

        # Iterar sobre records (no necesitamos el shape aquí, pero iterShapeRecords es conveniente)
        # Alternativamente, podrías usar sf.iterRecords() si solo necesitas los atributos
        for shapeRecord in sf.iterShapeRecords():
            record = shapeRecord.record

            # Extraer los atributos necesarios
            indice_centralidad = record[indice_id]
            nombre_centralidad = record[indice_nombre]
            tipo_centralidad = record[indice_tipo]

            # Escribir una fila por cada registro/polígono
            writer2.writerow([indice_centralidad, nombre_centralidad, tipo_centralidad])

    print("CSV 2 creado exitosamente.")


except shapefile.ShapefileException as e:
    print(f"Error al abrir o leer el Shapefile: {e}")
except FileNotFoundError:
    print(f"Error: No se encontró el archivo Shapefile en {ruta_shapefile}")
except IndexError:
    print(f"Error: Índice de campo inválido al acceder a los registros.")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```
